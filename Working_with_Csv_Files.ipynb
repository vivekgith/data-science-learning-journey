# 3. Opening a csv file from URl

```python
import requests
import pandas as pd
from io import StringIO  # Corrected from StringIo to StringIO

url = "https://people.sc.fsu.edu/~jburkardt/data/csv/addresses.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}  # Fixed string literal

req = requests.get(url, headers=headers)
data = StringIO(req.text)
pd.read_csv(data)  # Corrected from pd. read_sv to pd.read_csv
```
