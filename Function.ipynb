# Takes Nothing , Return Something

```python
def add3():
    print("Enter two Numbers")
    a, b = int(input()), int(input())
    c = a + b
    return c
def main():
    s = add3()
    print("Result is", s)
main()    
    
    
```

# Take Something , Return Something

```python
def add4(a, b):
    c = a + b
    return c
def main():
    x, y = 10, 20
    s = add4(x, y)
    print("sum of", x, "and", y, "is", s)
main()    
```
