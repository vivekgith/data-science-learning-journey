# 2.Write a python script to print first 10 multiplies of N.


```python
N = int(input("Enter a number: "))

for i in range(1, 11):
    print(N * i)
```

# 3.Write a python Script to print first M multiplies of N.

```python
N = int(input("Enter the number (N): "))
M = int(input("How many multiples (M): "))

for i in range(1, M + 1):
    print(N * i)
```

# 4.Write a python Script to print first 10 Multiplies of N in reverse order.

```python
N = int(input("Enter a number: "))

for i in range(10, 0, -1):
    print(N * i)
```

# 5.Write a python script to print table of user’s choice.

```python
N = int(input("Enter the number for table: "))

for i in range(1, 11):
    print(N, "x", i, "=", N * i)
```
