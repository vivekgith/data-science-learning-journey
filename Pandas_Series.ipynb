**Pandas Series**

```python
import pandas as pd
x=[1,2,7,9]
x1=pd.Series(x)
print(x1)
```

**Labeling or Accessing value via index**

**Accessing element via Indexing**

```python
print(x1[0])
print(x1[1])
print(x1[2])
```

**Changing The Index**

```python
x1=pd.Series(x,index=["x","y","z","a"])
print(x1)
#Accessing element via changing index.
print(x1['x'])
print(x1['y'])
print(x1['z'])
print(x1['a'])
```

**Taking Dictionary as series**

```python
cal={"day1":420,"day2":380,"day3":390}
dict_series=pd.Series(cal)
print(dict_series)
```

**Want get only two element from dict**

```python
dict_series=pd.Series(cal,index=['day1','day2'])
print(dict_series)
```

**DataFrame Introduction**

```python
x={"cal":[420,380,390],"duration":[50,40,45]}
x1=pd.DataFrame(x)
print(x1)
#DataFrame is the Whole Table.
#Series is like a column.
```
